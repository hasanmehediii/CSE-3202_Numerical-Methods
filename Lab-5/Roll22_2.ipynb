```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f2(t, y):
    return y * (np.sin(t))**3

def exact2(t):
    return np.exp(-np.cos(t) + ((np.cos(t))**3)/3 + 2/3)
```

```python
def euler_method(f, t0, y0, h, tend):
    n = int(round((tend - t0) / h))
    t = np.zeros(n+1)
    y = np.zeros(n+1)
    t[0] = t0
    y[0] = y0
    for i in range(n):
        y[i+1] = y[i] + h * f(t[i], y[i])
        t[i+1] = t[i] + h
    return t, y
```

```python
def heun_method(f, t0, y0, h, tend):
    n = int(round((tend - t0) / h))
    t = np.zeros(n+1)
    y = np.zeros(n+1)
    t[0] = t0
    y[0] = y0
    for i in range(n):
        k1 = f(t[i], y[i])
        y_pred = y[i] + h * k1
        k2 = f(t[i] + h, y_pred)
        y[i+1] = y[i] + (h / 2.0) * (k1 + k2)
        t[i+1] = t[i] + h
    return t, y
```

```python
def midpoint_method(f, t0, y0, t_end, h):
    steps = int(round((t_end - t0) / h))
    t_values = np.zeros(steps + 1)
    y_values = np.zeros(steps + 1)
    t_values[0] = t0
    y_values[0] = y0
    
    for i in range(steps):
        t = t_values[i]
        y = y_values[i]
        
        k1 = f(t, y)
        t_mid = t + h / 2
        y_mid = y + (h / 2) * k1
        k2 = f(t_mid, y_mid)
        
        y_values[i+1] = y + h * k2
        t_values[i+1] = t + h
        
    return t_values, y_values
```

```python
h_list2 = [0.1, 0.005, 0.001, 0.0005, 0.0001]
t0b, y0b, tendb = 0, 1, 5

euler_results2 = {}
heun_results2 = {}
midpoint_results2 = {}

for h in h_list2:
    te, ye = euler_method(f2, t0b, y0b, h, tendb)
    th, yh = heun_method(f2, t0b, y0b, h, tendb)
    tm, ym = midpoint_method(f2, t0b, y0b, h, tendb)

    yexact_e = exact2(te)
    yexact_h = exact2(th)
    yexact_m = exact2(tm)

    err_e = np.abs(yexact_e - ye)
    err_h = np.abs(yexact_h - yh)
    err_m = np.abs(yexact_m - ym)

    euler_results2[h] = (te, ye, yexact_e, err_e)
    heun_results2[h] = (th, yh, yexact_h, err_h)
    midpoint_results2[h] = (tm, ym, yexact_m, err_m)
```

```python
for h in h_list2:
    te, ye, yexact_e, err_e = euler_results2[h]
    print(f"\n--- Euler Method Table (h = {h}) ---")
    print(f"{'t':>8}{'Euler y':>15}{'Exact y':>15}{'Error':>15}")
    if len(te) <= 20:
        for i in range(len(te)):
            print(f"{te[i]:8.4f}{ye[i]:15.6f}{yexact_e[i]:15.6f}{err_e[i]:15.6f}")
    else:
        for i in range(10):
            print(f"{te[i]:8.4f}{ye[i]:15.6f}{yexact_e[i]:15.6f}{err_e[i]:15.6f}")
        print("...")
        for i in range(len(te)-10, len(te)):
            print(f"{te[i]:8.4f}{ye[i]:15.6f}{yexact_e[i]:15.6f}{err_e[i]:15.6f}")

for h in h_list2:
    th, yh, yexact_h, err_h = heun_results2[h]
    print(f"\n--- Heun Method Table (h = {h}) ---")
    print(f"{'t':>8}{'Heun y':>15}{'Exact y':>15}{'Error':>15}")
    if len(th) <= 20:
        for i in range(len(th)):
            print(f"{th[i]:8.4f}{yh[i]:15.6f}{yexact_h[i]:15.6f}{err_h[i]:15.6f}")
    else:
        for i in range(10):
            print(f"{th[i]:8.4f}{yh[i]:15.6f}{yexact_h[i]:15.6f}{err_h[i]:15.6f}")
        print("...")
        for i in range(len(th)-10, len(th)):
            print(f"{th[i]:8.4f}{yh[i]:15.6f}{yexact_h[i]:15.6f}{err_h[i]:15.6f}")


for h in h_list2:
    tm, ym, yexact_m, err_m = midpoint_results2[h]
    print(f"\n--- Midpoint Method Table (h = {h}) ---")
    print(f"{'t':>8}{'Midpoint y':>15}{'Exact y':>15}{'Error':>15}")
    if len(tm) <= 20:
        for i in range(len(tm)):
            print(f"{tm[i]:8.4f}{ym[i]:15.6f}{yexact_m[i]:15.6f}{err_m[i]:15.6f}")
    else:
        for i in range(10):
            print(f"{tm[i]:8.4f}{ym[i]:15.6f}{yexact_m[i]:15.6f}{err_m[i]:15.6f}")
        print("...")
        for i in range(len(tm)-10, len(tm)):
            print(f"{tm[i]:8.4f}{ym[i]:15.6f}{yexact_m[i]:15.6f}{err_m[i]:15.6f}")
```

```python
plt.figure(figsize=(10, 7))
tt2 = np.linspace(t0b, tendb, 1000)
plt.plot(tt2, exact2(tt2), 'k-', linewidth=2, label='Exact Solution')

for h in h_list2:
    te, ye, _, _ = euler_results2[h]
    plt.plot(te, ye, '--', label=f'Euler h={h}')

for h in h_list2:
    th, yh, _, _ = heun_results2[h]
    plt.plot(th, yh, ':', label=f'Heun h={h}')

for h in h_list2:
    tm, ym, _, _ = midpoint_results2[h]
    plt.plot(tm, ym, '-.', label=f'Midpoint h={h}')

plt.xlabel('t')
plt.ylabel('y')
plt.title('Task 2: Exact vs Euler vs Heun vs Midpoint')
# plt.xlim(1, 2)
# plt.ylim(1, 2)
plt.legend()
plt.grid(True)
plt.show()
```
